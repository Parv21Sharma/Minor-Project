```python
import cv2
import os
video_folder = r"C:\Users\Hp\Desktop\fightvideos"
output_folder = r"C:\Users\Hp\Desktop\rawphotosfordanger"
max_frames_per_video = 5
video_number = 1

for video_file in os.listdir(video_folder):
    video_path = os.path.join(video_folder, video_file)
    
    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    if total_frames == 0:
        continue
    frame_positions = [
        int(total_frames * i / (max_frames_per_video + 1))
        for i in range(1, max_frames_per_video + 1)
    ]
    saved_count = 0
    current_frame = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        if current_frame in frame_positions:
            filename = f"danger_video{video_number}_{saved_count}.jpg"
            cv2.imwrite(os.path.join(output_folder, filename), frame)
            saved_count += 1
        
        current_frame += 1
    
    cap.release()
    video_number += 1

print("Extraction complete.")
```
